```python
# ------------------------------------------------------------
# 1. IMPORT LIBRARIES
# ------------------------------------------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.model_selection import StratifiedKFold
from sklearn.model_selection import GridSearchCV

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import AdaBoostClassifier

from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
    roc_auc_score,
    roc_curve,
    classification_report
)
```

```python
# ------------------------------------------------------------
# 2. LOAD THE DATASET
# ------------------------------------------------------------

df = pd.read_csv("UCI_Credit_Card.csv")

print("Dataset loaded successfully!")
```

```python
# ------------------------------------------------------------
# 3. BASIC DATA EXPLORATION
# ------------------------------------------------------------

print("\nFirst 5 rows:")
print(df.head())

print("\nDataset shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
print(df.info())

print("\nStatistical summary:")
print(df.describe())
```

```python
# ------------------------------------------------------------
# 4. CHECK MISSING VALUES
# ------------------------------------------------------------

print("\nMissing values:")
print(df.isnull().sum())

print("\nTotal missing values:")
print(df.isnull().sum().sum())
```

```python
# ------------------------------------------------------------
# 5. CHECK DUPLICATE ROWS
# ------------------------------------------------------------

print("\nNumber of duplicate rows:")
print(df.duplicated().sum())

# Remove duplicate rows if any
df = df.drop_duplicates()

print("\nShape after removing duplicates:")
print(df.shape)
```

```python

# ------------------------------------------------------------
# 6. CHECK TARGET VARIABLE
# ------------------------------------------------------------

target = "default.payment.next.month"

print("\nTarget value counts:")
print(df[target].value_counts())

print("\nTarget percentages:")
print(df[target].value_counts(normalize=True) * 100)
```

```python
# ------------------------------------------------------------
# 7. EDA - TARGET DISTRIBUTION
# ------------------------------------------------------------

df[target].value_counts().plot(kind="bar")

plt.title("Credit Card Default Distribution")
plt.xlabel("Default Next Month")
plt.ylabel("Number of Customers")
plt.xticks([0, 1], ["No Default", "Default"], rotation=0)

plt.show()
```

```python
# ------------------------------------------------------------
# 8. CHECK UNIQUE VALUES
# ------------------------------------------------------------

print("\nNumber of unique values in each column:")
print(df.nunique())
```

```python
# ------------------------------------------------------------
# 9. REMOVE ID COLUMN
# ------------------------------------------------------------

# ID is only an identifier and does not represent
# meaningful customer information.

if "ID" in df.columns:
    df = df.drop("ID", axis=1)

print("\nShape after removing ID:")
print(df.shape)
```

```python
# ------------------------------------------------------------
# 10. FEATURE ENGINEERING
# ------------------------------------------------------------

# Total bill amount over the previous 6 months

bill_columns = [
    "BILL_AMT1",
    "BILL_AMT2",
    "BILL_AMT3",
    "BILL_AMT4",
    "BILL_AMT5",
    "BILL_AMT6"
]

# Total payment amount over the previous 6 months

payment_columns = [
    "PAY_AMT1",
    "PAY_AMT2",
    "PAY_AMT3",
    "PAY_AMT4",
    "PAY_AMT5",
    "PAY_AMT6"
]

df["TOTAL_BILL_AMT"] = df[bill_columns].sum(axis=1)

df["TOTAL_PAY_AMT"] = df[payment_columns].sum(axis=1)

df["AVG_BILL_AMT"] = df[bill_columns].mean(axis=1)

df["AVG_PAY_AMT"] = df[payment_columns].mean(axis=1)


print("\nDataset after feature engineering:")
print(df.head())

print("\nNew shape:")
print(df.shape)

```

```python
# ------------------------------------------------------------
# 11. CHECK CORRELATION WITH TARGET
# ------------------------------------------------------------

correlation = df.corr(numeric_only=True)[target].sort_values(
    ascending=False
)

print("\nCorrelation with target:")
print(correlation)
```

```python
# ------------------------------------------------------------
# 12. SEPARATE FEATURES AND TARGET
# ------------------------------------------------------------

X = df.drop(target, axis=1)

y = df[target]

print("\nFeature shape:")
print(X.shape)

print("\nTarget shape:")
print(y.shape)
```

```python
# ------------------------------------------------------------
# 13. CHECK MISSING VALUES IN FEATURES
# ------------------------------------------------------------

print("\nMissing values in X:")
print(X.isnull().sum().sum())
```

```python
# ------------------------------------------------------------
# 14. TRAIN-TEST SPLIT
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining data shape:")
print(X_train.shape)

print("\nTesting data shape:")
print(X_test.shape)
```

```python
# ------------------------------------------------------------
# 15. CREATE BASE ESTIMATOR
# ------------------------------------------------------------

# Decision stump = Decision Tree with only one level

base_tree = DecisionTreeClassifier(
    max_depth=1,
    random_state=42
)
```

```python
# ------------------------------------------------------------
# 16. CREATE ADABOOST MODEL
# ------------------------------------------------------------

ada_model = AdaBoostClassifier(
    estimator=base_tree,
    random_state=42
)
```

```python
# ------------------------------------------------------------
# 17. CREATE 10-FOLD CROSS VALIDATION
# ------------------------------------------------------------

cv = StratifiedKFold(
    n_splits=10,
    shuffle=True,
    random_state=42
)
```

```python

# ------------------------------------------------------------
# 18. DEFINE HYPERPARAMETER GRID
# ------------------------------------------------------------

param_grid = {
    "n_estimators": [50, 100, 150, 200],
    "learning_rate": [0.01, 0.05, 0.1, 0.5, 1.0]
}
```

```python
# ------------------------------------------------------------
# 19. GRID SEARCH WITH 10-FOLD CROSS VALIDATION
# ------------------------------------------------------------

grid_search = GridSearchCV(
    estimator=ada_model,
    param_grid=param_grid,
    cv=cv,
    scoring="roc_auc",
    n_jobs=-1,
    verbose=1
)
```

```python
# ------------------------------------------------------------
# 20. TRAIN AND TUNE THE MODEL
# ------------------------------------------------------------

grid_search.fit(X_train, y_train)

```

```python
# ------------------------------------------------------------
# 21. DISPLAY BEST PARAMETERS
# ------------------------------------------------------------

print("\nBest Parameters:")
print(grid_search.best_params_)

print("\nBest Cross-Validation ROC-AUC:")
print(grid_search.best_score_)
```

```python
# ------------------------------------------------------------
# 22. GET BEST MODEL
# ------------------------------------------------------------

best_model = grid_search.best_estimator_

print("\nBest AdaBoost Model:")
print(best_model)
```

```python
# ------------------------------------------------------------
# 23. MAKE PREDICTIONS
# ------------------------------------------------------------

y_pred = best_model.predict(X_test)

# Probability of class 1 (default)
y_prob = best_model.predict_proba(X_test)[:, 1]
```

```python
# ------------------------------------------------------------
# 24. ACCURACY
# ------------------------------------------------------------

accuracy = accuracy_score(y_test, y_pred)

print("\nAccuracy:")
print(accuracy)
```

```python
# ------------------------------------------------------------
# 25. CONFUSION MATRIX
# ------------------------------------------------------------

cm = confusion_matrix(y_test, y_pred)

print("\nConfusion Matrix:")
print(cm)

ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["No Default", "Default"]
).plot()

plt.title("AdaBoost Confusion Matrix")

plt.show()
```

```python
# ------------------------------------------------------------
# 26. ROC-AUC
# ------------------------------------------------------------

roc_auc = roc_auc_score(y_test, y_prob)

print("\nROC-AUC:")
print(roc_auc)
```

```python
# ------------------------------------------------------------
# 27. ROC CURVE
# ------------------------------------------------------------

fpr, tpr, thresholds = roc_curve(y_test, y_prob)

plt.plot(
    fpr,
    tpr,
    label="AdaBoost"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.xlabel("False Positive Rate")

plt.ylabel("True Positive Rate")

plt.title("ROC Curve")

plt.legend()

plt.show()


# ------------------------------------------------------------
# 28. CLASSIFICATION REPORT
# ------------------------------------------------------------

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```
